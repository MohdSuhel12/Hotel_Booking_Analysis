```python
df["total_guests"] = (
    df["adults"]
    + df["children"]
    + df["babies"]
)
```

```python
df["estimated_revenue"] = (
    df["adr"] * df["total_nights"]
)
```

```python
df["cancellation_status"] = df["is_canceled"].map({
    0: "Not Cancelled",
    1: "Cancelled"
})
```

```python
print("Total bookings:", len(df))

cancellation_rate = df["is_canceled"].mean() * 100

print("Cancellation rate:", round(cancellation_rate, 2), "%")
```

```python
hotel_bookings = df["hotel"].value_counts()

print(hotel_bookings)
```

```python
hotel_cancellation = (
    df.groupby("hotel")["is_canceled"]
    .mean()
    .mul(100)
    .round(2)
)

print(hotel_cancellation)
```

```python
average_adr = df["adr"].mean()

print("Average Daily Rate:", round(average_adr, 2))
```

```python
hotel_adr = (
    df.groupby("hotel")["adr"]
    .mean()
    .round(2)
)

print(hotel_adr)
```

```python
segment_analysis = (
    df.groupby("market_segment")
    .agg(
        total_bookings=("hotel", "count"),
        cancellation_rate=("is_canceled", "mean"),
        average_adr=("adr", "mean")
    )
    .reset_index()
)

segment_analysis["cancellation_rate"] = (
    segment_analysis["cancellation_rate"] * 100
).round(2)

segment_analysis["average_adr"] = (
    segment_analysis["average_adr"].round(2)
)

segment_analysis.sort_values(
    "total_bookings",
    ascending=False
)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="cancellation_status"
)

plt.title("Booking Cancellation Distribution")
plt.xlabel("Booking Status")
plt.ylabel("Number of Bookings")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="hotel",
    y="is_canceled"
)

plt.title("Cancellation Rate by Hotel")
plt.xlabel("Hotel")
plt.ylabel("Cancellation Rate")
plt.show()
```

```python
month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

monthly_bookings = (
    df.groupby("arrival_date_month")
    .size()
    .reindex(month_order)
)

print(monthly_bookings)
```

```python
plt.figure(figsize=(12, 5))

monthly_bookings.plot(
    kind="line",
    marker="o"
)

plt.title("Monthly Booking Trend")
plt.xlabel("Month")
plt.ylabel("Number of Bookings")
plt.xticks(rotation=45)
plt.show()
```

```python
df["lead_time_group"] = pd.cut(
    df["lead_time"],
    bins=[-1, 30, 90, 180, float("inf")],
    labels=["0-30 Days", "31-90 Days", "91-180 Days", "180+ Days"]
)
```

```python
lead_time_analysis = (
    df.groupby("lead_time_group", observed=False)
    .agg(
        total_bookings=("hotel", "count"),
        cancellation_rate=("is_canceled", "mean")
    )
    .reset_index()
)

lead_time_analysis["cancellation_rate"] = (
    lead_time_analysis["cancellation_rate"] * 100
).round(2)

print(lead_time_analysis)
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=lead_time_analysis,
    x="lead_time_group",
    y="cancellation_rate"
)

plt.title("Cancellation Rate by Lead Time")
plt.xlabel("Booking Lead Time")
plt.ylabel("Cancellation Rate (%)")
plt.xticks(rotation=20)
plt.show()
```

```python
deposit_analysis = (
    df.groupby("deposit_type")
    .agg(
        total_bookings=("hotel", "count"),
        cancellation_rate=("is_canceled", "mean")
    )
    .reset_index()
)

deposit_analysis["cancellation_rate"] = (
    deposit_analysis["cancellation_rate"] * 100
).round(2)

print(deposit_analysis)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=deposit_analysis,
    x="deposit_type",
    y="cancellation_rate"
)

plt.title("Cancellation Rate by Deposit Type")
plt.xlabel("Deposit Type")
plt.ylabel("Cancellation Rate (%)")
plt.xticks(rotation=20)
plt.show()
```

```python
customer_analysis = (
    df.groupby("customer_type")
    .agg(
        total_bookings=("hotel", "count"),
        cancellation_rate=("is_canceled", "mean"),
        average_adr=("adr", "mean")
    )
    .reset_index()
)

customer_analysis["cancellation_rate"] = (
    customer_analysis["cancellation_rate"] * 100
).round(2)

customer_analysis["average_adr"] = (
    customer_analysis["average_adr"].round(2)
)

print(customer_analysis)
```

```python
top_countries = (
    df["country"]
    .value_counts()
    .head(10)
    .reset_index()
)

top_countries.columns = ["country", "bookings"]

print(top_countries)
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    data=top_countries,
    x="country",
    y="bookings"
)

plt.title("Top 10 Customer Countries")
plt.xlabel("Country")
plt.ylabel("Bookings")
plt.xticks(rotation=45)
plt.show()
```

```python
completed_bookings = df[df["is_canceled"] == 0].copy()
```

```python
total_revenue = completed_bookings["estimated_revenue"].sum()

print("Estimated Revenue:", round(total_revenue, 2))
```

```python
hotel_revenue = (
    completed_bookings
    .groupby("hotel")["estimated_revenue"]
    .sum()
    .sort_values(ascending=False)
)

print(hotel_revenue)
```

```python
plt.figure(figsize=(8, 5))

hotel_revenue.plot(
    kind="bar",
    color=["steelblue", "orange"]
)

plt.title("Estimated Revenue by Hotel")
plt.xlabel("Hotel")
plt.ylabel("Estimated Revenue")
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

hotel_revenue.plot(
    kind="bar",
    color=["steelblue", "orange"]
)

plt.title("Estimated Revenue by Hotel")
plt.xlabel("Hotel")
plt.ylabel("Estimated Revenue")
plt.xticks(rotation=0)
plt.show()
```
